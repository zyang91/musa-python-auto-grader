# Assignment #1: The Donut Effect for Philadelphia ZIP Codes

## 1. Load the data

```python
import os

import pandas as pd

# Falls back to where the file lives on my laptop
path = "data/zillow_zhvi.csv"
if not os.path.exists(path):
    path = "/Users/student/Desktop/musa/zillow_zhvi.csv"
df = pd.read_csv(path)
df.shape
```

## 2. Trim the data to just Philadelphia

```python
philly = df[df["City"] == "Philadelphia"]
philly.shape
```

## 3. Melt the data into tidy format

```python
dates = [c for c in philly.columns if c[:2] in ('20', '19')]
tidy = philly.melt(
    id_vars=['RegionName'],
    value_vars=dates,
    var_name='date',
    value_name='value',
)
tidy.shape
```

## 4. Split the data for ZIP codes in/outside Center City

```python
# I only used the four main Center City ZIP codes
greater_center_city_zip_codes = [19102, 19103, 19106, 19107]
cc = tidy[tidy['RegionName'].isin(greater_center_city_zip_codes)]
non_cc = tidy[~tidy['RegionName'].isin(greater_center_city_zip_codes)]
cc['RegionName'].unique()
```

## 5. Compare home value appreciation in Philadelphia

```python
def calculate_percent_increase(group_df):
    """Percent increase over the period."""
    first = group_df['value'].iloc[0]
    last = group_df['value'].iloc[-1]
    return (last - first) / first * 100
```

```python
cc_change = cc.groupby('RegionName').apply(calculate_percent_increase)
non_cc_change = non_cc.groupby('RegionName').apply(calculate_percent_increase)
cc_avg = cc_change.mean()
non_cc_avg = non_cc_change.mean()
print(cc_avg, non_cc_avg)
```
